## 💻 PHẦN A: DEMO "VÀI ĐƯỜNG CƠ BẢN" TRÊN GOOGLE COLAB (CÙNG GIẢNG VIÊN)
👉 *Nhấn `Shift + Enter` hoặc bấm nút **Play (▶️)** ở góc trái từng ô để chạy code.*

```python
# [ĐƯỜNG CƠ BẢN 1]: Khởi tạo Vector (x ∈ R^n) & Ma trận (X ∈ R^(m×n)) bằng NumPy
import numpy as np

# 1. Tạo Vector x ∈ R^4 (Hồ sơ 1 sinh viên: [GPA, Số_Project, Điểm_Code, Tiếng_Anh])
x = np.array([3.5, 4, 8.0, 7.0])
print("Vector x:", x)
print("Kích thước của x (shape):", x.shape)

# 2. Tạo Ma trận X ∈ R^(3x4) (Bảng gồm 3 sinh viên, mỗi người 4 thông số)
X = np.array([
    [3.5, 4, 8.0, 7.0],
    [2.8, 2, 6.0, 5.5],
    [3.9, 6, 9.5, 8.5]
])
print("\nKích thước Ma trận X (dòng, cột):", X.shape)
```

```python
# [ĐƯỜNG CƠ BẢN 2]: Tính Tổng Sigma (∑) và Trung bình cộng (x̄)
diem_kiem_tra = np.array([8.0, 7.0, 9.0, 6.0])

tong_sigma = np.sum(diem_kiem_tra)    # Tương đương ∑ x_i
trung_binh = np.mean(diem_kiem_tra)   # Tương đương (1/n) ∑ x_i
diem_cao_nhat = np.max(diem_kiem_tra)

print("Tổng Sigma (∑ x_i)       =", tong_sigma)
print("Trung bình cộng (x̄)      =", trung_binh)
print("Điểm cao nhất trong mảng =", diem_cao_nhat)
```

```python
# [ĐƯỜNG CƠ BẢN 3]: Code công thức 1 Nơ-ron Nhân tạo (ŷ = w·x + b)
# Điểm thành phần của sinh viên: [Chuyên cần, Giữa kỳ, Cuối kỳ]
x = np.array([9.0, 7.5, 8.0])

# Trọng số môn học: [10%, 30%, 60%] và điểm cộng ưu tiên b = 0.2
w = np.array([0.1, 0.3, 0.6])
b = 0.2

# Tính đầu ra của Nơ-ron bằng tích vô hướng np.dot
y_hat = np.dot(w, x) + b
print(f"Điểm tổng kết dự đoán (ŷ) = {y_hat:.2f} điểm")
```

```python
# [ĐƯỜNG CƠ BẢN 4]: Sức mạnh Ma trận — Chấm điểm cả danh sách không cần vòng lặp for!
X = np.array([
    [3.5, 4, 8.0, 7.0], # Ứng viên 1
    [2.8, 2, 6.0, 5.5], # Ứng viên 2
    [3.9, 6, 9.5, 8.5]  # Ứng viên 3
])

w = np.array([1.5, 1.5, 2.0, 1.5])
b = 2.0

diem_ung_vien = np.dot(X, w) + b
print("Điểm của 3 ứng viên lần lượt là:", diem_ung_vien)
```

## 📝 PHẦN B: BÀI TẬP THỰC HÀNH TẠI LỚP (DÀNH CHO SINH VIÊN)
👉 *Hãy sửa các chỗ trống hoặc thay số liệu của bạn vào 3 bài tập dưới đây rồi bấm chạy!*

```python
# ==============================================================================
# BÀI TẬP 1 (Mức Cơ bản - 10 phút): Số hóa Bảng điểm Cá nhân & Xét Học bổng
# Yêu cầu: Thay 5 điểm môn học của bạn vào mảng, tính tổng ∑, trung bình x̄ và max
# ==============================================================================
import numpy as np

# 1. Thay điểm 5 môn học của bạn vào đây:
diem_5_mon = np.array([8.5, 7.5, 9.0, 8.0, 8.5])

# 2. Kiểm tra kích thước (.shape)
print("Kích thước vector điểm:", diem_5_mon.shape)

# 3. Tính Tổng Sigma (∑), Trung bình (x̄) và Điểm cao nhất
tong_diem = np.sum(diem_5_mon)
diem_tb   = np.mean(diem_5_mon)
diem_max  = np.max(diem_5_mon)

print("Tổng điểm 5 môn (∑):", tong_diem)
print("Điểm trung bình (x̄):", round(diem_tb, 2))
print("Môn điểm cao nhất  :", diem_max)

if diem_tb >= 8.0:
    print("🎉 Kết quả: Đạt Học bổng Khuyến khích!")
else:
    print("💪 Kết quả: Cố gắng thêm ở học kỳ tới nhé!")
```

```python
# ==============================================================================
# BÀI TẬP 2 (Mức Vận dụng - 15 phút): "Cứu sáng" cho Bức Ảnh Bị Tối bằng Ma trận
# Yêu cầu: Nhân độ tương phản 1.5 lần, cộng thêm 60 độ sáng và vẽ ảnh đối chiếu
# ==============================================================================
import numpy as np
import matplotlib.pyplot as plt

anh_toi = np.array([
    [10,  20,  30,  15],
    [40, 110, 120,  35],
    [45, 130, 100,  25],
    [10,  15,  20,  10]
])

# 1. Tăng tương phản 1.5 lần và cộng 60 độ sáng (giới hạn trong [0, 255])
anh_sang = np.clip(anh_toi * 1.5 + 60, 0, 255)

# 2. Tách nền nhị phân (>= 128 thành 255, ngược lại thành 0)
anh_nhi_phan = np.where(anh_sang >= 128, 255, 0)

print("Độ sáng TB ban đầu:", np.mean(anh_toi))
print("Độ sáng TB sau khi cứu sáng:", np.mean(anh_sang))

# Hiển thị trực quan 3 ma trận thành hình ảnh
fig, axes = plt.subplots(1, 3, figsize=(9, 3))
axes[0].imshow(anh_toi, cmap='gray', vmin=0, vmax=255)
axes[0].set_title("1. Ảnh gốc bị tối")
axes[1].imshow(anh_sang, cmap='gray', vmin=0, vmax=255)
axes[1].set_title("2. Ảnh sau cứu sáng")
axes[2].imshow(anh_nhi_phan, cmap='gray', vmin=0, vmax=255)
axes[2].set_title("3. Ảnh nhị phân (0/255)")
plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# BÀI TẬP 3 (Mini-Project Cuối Buổi - 20 phút):
# Tạo Web App AI Chấm điểm CV & Dự đoán Lương Thực tập IT bằng Gradio
# ==============================================================================
!pip install -q gradio

import numpy as np
import gradio as gr

def cham_diem_cv(ho_ten, gpa, so_project, diem_code, tieng_anh, loai_cong_ty):
    # 1. Biểu diễn hồ sơ ứng viên thành Vector x ∈ R^4
    x = np.array([gpa, so_project, diem_code, tieng_anh], dtype=float)
    
    # 2. Chọn Vector trọng số w ∈ R^4 theo loại công ty
    if loai_cong_ty == "Công ty Product / AI Startup":
        w = np.array([1.0, 1.8, 2.5, 1.2])
        b = 2.0
    elif loai_cong_ty == "Tập đoàn Đa quốc gia (Global)":
        w = np.array([2.2, 1.2, 1.5, 2.5])
        b = 1.5
    else:
        w = np.array([1.5, 1.5, 2.0, 1.5])
        b = 2.0
        
    # 3. Tính điểm bằng công thức Nơ-ron nhân tạo: y_hat = w · x + b
    diem_tho = float(np.dot(w, x) + b)
    diem_100 = min(100, round((diem_tho / 52.0) * 100))
    luong_du_kien = round(diem_tho * 0.32, 1)
    
    if diem_100 >= 72:
        ket_qua = "🔥 ĐẬU OFFER THỰC TẬP / FRESHER"
    elif diem_100 >= 55:
        ket_qua = "⚡ VÀO VÒNG PHỎNG VẤN KỸ THUẬT"
    else:
        ket_qua = "💡 CẦN BỔ SUNG THÊM PROJECT GITHUB"
        
    return f"""### 🎯 KẾT QUẢ ĐÁNH GIÁ HỒ SƠ: {ho_ten.upper()}
- **Trạng thái:** {ket_qua}
- **Điểm Nơ-ron (`np.dot(w, x) + b`):** `{diem_tho:.2f}` điểm (Quy đổi: **{diem_100}/100**)
- **Mức lương Thực tập / Fresher đề xuất:** **{luong_du_kien} triệu VNĐ / tháng**
---
#### 📐 Giải mã Toán học bên dưới:
- **Vector ứng viên `x`:** `{x.tolist()}`
- **Vector trọng số công ty `w`:** `{w.tolist()}` (Bias `b = {b}`)
- **Phép tính:** `({w[0]}×{gpa}) + ({w[1]}×{so_project}) + ({w[2]}×{diem_code}) + ({w[3]}×{tieng_anh}) + {b} = {diem_tho:.2f}`
"""

demo = gr.Interface(
    fn=cham_diem_cv,
    inputs=[
        gr.Textbox(label="Họ và tên ứng viên", value="Nguyễn Văn An"),
        gr.Slider(0.0, 4.0, value=3.2, step=0.1, label="Điểm GPA Đại học (Thang 4.0)"),
        gr.Slider(0, 15, value=4, step=1, label="Số Project thực tế trên GitHub"),
        gr.Slider(0.0, 10.0, value=8.0, step=0.5, label="Điểm Code / Thuật toán (Thang 10)"),
        gr.Slider(0.0, 10.0, value=7.0, step=0.5, label="Kỹ năng Tiếng Anh (Thang 10)"),
        gr.Radio(
            ["Công ty Product / AI Startup", "Tập đoàn Đa quốc gia (Global)", "Chương trình Thực tập sinh Tài năng"],
            value="Công ty Product / AI Startup",
            label="Chọn Gu Tuyển Dụng của Công ty"
        )
    ],
    outputs=gr.Markdown(label="Kết quả Phân tích từ Mô hình"),
    title="🚀 Sản phẩm Bài tập 3: Hệ thống AI Chấm điểm CV & Dự đoán Lương IT",
    description="Hoàn thành bởi Sinh viên trong Buổi 1 — Khóa học Toán AI Thực Chiến 2026."
)

demo.launch(share=True)
```
