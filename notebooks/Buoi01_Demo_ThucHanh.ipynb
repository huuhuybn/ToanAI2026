```python
#@title 1️⃣ THÍ NGHIỆM 1: Dịch Ký hiệu Toán (Vector, Tổng Sigma & Nơ-ron) sang NumPy { run: "auto" }
#@markdown 👉 **Hướng dẫn:** Hãy kéo các thanh trượt bên dưới để nhập điểm của bạn. Code sẽ tự động chạy và dịch sang công thức Toán học!

import numpy as np

diem_chuyen_can = 9.0 #@param {type:"slider", min:0, max:10, step:0.5}
diem_giua_ky = 7.5 #@param {type:"slider", min:0, max:10, step:0.5}
diem_cuoi_ky = 8.0 #@param {type:"slider", min:0, max:10, step:0.5}
diem_cong_bias = 0.5 #@param {type:"slider", min:0, max:2, step:0.1}

# 1. Biểu diễn điểm của bạn thành Vector x ∈ R^3 (Mảng 1 chiều)
x = np.array([diem_chuyen_can, diem_giua_ky, diem_cuoi_ky])

# 2. Trọng số môn học w ∈ R^3: [10%, 30%, 60%]
w = np.array([0.1, 0.3, 0.6])

# 3. Tính Tổng Sigma (∑ x_i) và Trung bình cộng (x̄)
tong_sigma = np.sum(x)
trung_binh = np.mean(x)

# 4. Tính điểm tổng kết bằng công thức 1 Nơ-ron: y_hat = w · x + b
diem_tong_ket = np.dot(w, x) + diem_cong_bias

print("="*65)
print(f"📌 Vector điểm của bạn (x ∈ R^3)       : {x} -> shape: {x.shape}")
print(f"📌 Tổng Sigma (∑ x_i = np.sum(x))      : {tong_sigma:.2f}")
print(f"📌 Trung bình ((1/n)∑ x_i = np.mean(x)): {trung_binh:.2f}")
print("-"*65)
print(f"🧠 Kết quả Nơ-ron (ŷ = w·x + b)        : {diem_tong_ket:.2f} điểm")
print("="*65)
```

```python
#@title 2️⃣ THÍ NGHIỆM 2: Máy tính nhìn Bức Ảnh bằng Ma trận số như thế nào? { run: "auto" }
#@markdown 👉 **Hướng dẫn:** Kéo thanh trượt `do_sang_b` và `tuong_phan_w` để xem phép cộng/nhân Ma trận làm thay đổi bức ảnh chữ số 8 ngay lập tức!

import numpy as np
import matplotlib.pyplot as plt

do_sang_b = 20 #@param {type:"slider", min:-150, max:150, step:10}
tuong_phan_w = 1.2 #@param {type:"slider", min:0.2, max:2.5, step:0.1}
dao_nguoc_mau = False #@param {type:"boolean"}

# Ma trận ảnh 8x8 biểu diễn chữ số 8 (Giá trị từ 0: Đen -> 255: Trắng)
img_goc = np.array([
    [0,  20, 200, 240, 240, 200,  20, 0],
    [0, 210, 250,  40,  40, 250, 210, 0],
    [0, 200, 240,  30,  30, 240, 200, 0],
    [0,  40, 220, 255, 255, 220,  40, 0],
    [0, 200, 240,  30,  30, 240, 200, 0],
    [0, 220, 250,  20,  20, 250, 220, 0],
    [0,  30, 210, 250, 250, 210,  30, 0],
    [0,   0,  20,  40,  40,  20,   0, 0]
], dtype=float)

# Áp dụng phép biến đổi tuyến tính lên toàn bộ Ma trận: Y = w * X + b
img_moi = np.clip(img_goc * tuong_phan_w + do_sang_b, 0, 255)
if dao_nguoc_mau:
    img_moi = 255 - img_moi

fig, axes = plt.subplots(1, 2, figsize=(8, 4))
axes[0].imshow(img_goc, cmap='gray', vmin=0, vmax=255)
axes[0].set_title(f"Ảnh Gốc (Mean = {np.mean(img_goc):.1f})")

axes[1].imshow(img_moi, cmap='gray', vmin=0, vmax=255)
axes[1].set_title(f"Sau biến đổi Ma trận (Mean = {np.mean(img_moi):.1f})")
plt.tight_layout()
plt.show()
```

```python
#@title 3️⃣ SẢN PHẨM CUỐI BUỔI: Chạy 1 Nút để Tạo Web App AI Chấm điểm CV & Dự đoán Lương IT
#@markdown 👉 **Hướng dẫn:** Bấm nút **Play (▶️)** ở góc trái ô này. Sau 10 giây, một **Giao diện Web App hoàn chỉnh** sẽ hiện ra ngay bên dưới kèm đường link `gradio.live` để bạn gửi cho bạn bè dùng thử!

!pip install -q gradio

import numpy as np
import gradio as gr

def cham_diem_cv(ho_ten, gpa, so_project, diem_code, tieng_anh, loai_cong_ty):
    # 1. Biểu diễn hồ sơ ứng viên thành Vector x ∈ R^4
    x = np.array([gpa, so_project, diem_code, tieng_anh], dtype=float)
    
    # 2. Chọn Vector trọng số w ∈ R^4 theo loại công ty
    if loai_cong_ty == "Công ty Product / AI Startup":
        w = np.array([1.0, 1.8, 2.5, 1.2])
        b = 2.0
    elif loai_cong_ty == "Tập đoàn Đa quốc gia (Global)":
        w = np.array([2.2, 1.2, 1.5, 2.5])
        b = 1.5
    else:
        w = np.array([1.5, 1.5, 2.0, 1.5])
        b = 2.0
        
    # 3. Tính điểm bằng công thức Nơ-ron nhân tạo: y_hat = w · x + b
    diem_tho = float(np.dot(w, x) + b)
    diem_100 = min(100, round((diem_tho / 52.0) * 100))
    luong_du_kien = round(diem_tho * 0.32, 1)
    
    if diem_100 >= 72:
        ket_qua = "🔥 ĐẬU OFFER THỰC TẬP / FRESHER"
    elif diem_100 >= 55:
        ket_qua = "⚡ VÀO VÒNG PHỎNG VẤN KỸ THUẬT"
    else:
        ket_qua = "💡 CẦN BỔ SUNG THÊM PROJECT GITHUB"
        
    bao_cao = f"""### 🎯 KẾT QUẢ ĐÁNH GIÁ HỒ SƠ: {ho_ten.upper()}
- **Trạng thái:** {ket_qua}
- **Điểm Nơ-ron (`np.dot(w, x) + b`):** `{diem_tho:.2f}` điểm (Quy đổi: **{diem_100}/100**)
- **Mức lương Thực tập / Fresher đề xuất:** **{luong_du_kien} triệu VNĐ / tháng**
---
#### 📐 Giải mã Toán học bên dưới lớp vỏ AI:
- **Vector ứng viên `x`:** `{x.tolist()}`
- **Vector trọng số công ty `w`:** `{w.tolist()}` (Bias `b = {b}`)
- **Phép tính:** `({w[0]}×{gpa}) + ({w[1]}×{so_project}) + ({w[2]}×{diem_code}) + ({w[3]}×{tieng_anh}) + {b} = {diem_tho:.2f}`
"""
    return bao_cao

demo = gr.Interface(
    fn=cham_diem_cv,
    inputs=[
        gr.Textbox(label="Họ và tên ứng viên", value="Nguyễn Văn An"),
        gr.Slider(0.0, 4.0, value=3.2, step=0.1, label="Điểm GPA Đại học (Thang 4.0)"),
        gr.Slider(0, 15, value=4, step=1, label="Số Project thực tế trên GitHub"),
        gr.Slider(0.0, 10.0, value=8.0, step=0.5, label="Điểm Code / Thuật toán (Thang 10)"),
        gr.Slider(0.0, 10.0, value=7.0, step=0.5, label="Kỹ năng Tiếng Anh (Thang 10)"),
        gr.Radio(
            ["Công ty Product / AI Startup", "Tập đoàn Đa quốc gia (Global)", "Chương trình Thực tập sinh Tài năng"],
            value="Công ty Product / AI Startup",
            label="Chọn Gu Tuyển Dụng của Công ty"
        )
    ],
    outputs=gr.Markdown(label="Kết quả Phân tích từ Mô hình"),
    title="🚀 Mini-App Buổi 1: Hệ thống AI Chấm điểm CV & Dự đoán Lương IT",
    description="Sản phẩm thực hành Buổi 1 — Khóa học Toán AI Thực Chiến dành cho Sinh viên CNTT."
)

demo.launch(share=True)
```
